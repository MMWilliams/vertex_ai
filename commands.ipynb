### Deleting all indices and all endpoints

```python
import subprocess

def run_command(command):
    try:
        result = subprocess.run(command, shell=True, check=True, capture_output=True, text=True)
        print("Output:", result.stdout)
        return result.stdout
    except subprocess.CalledProcessError as e:
        print("Error:", e.stderr)
        return None

# Delete all indexes
list_indexes_cmd = "gcloud ai indexes list --region=us-central1 --format='value(name)'"
indexes = run_command(list_indexes_cmd)

if indexes:
    for index in indexes.strip().split('\n'):
        if index:  # Check if index is not empty
            delete_cmd = f"gcloud ai indexes delete {index} --region=us-central1 --quiet"
            print(f"Deleting index: {index}")
            run_command(delete_cmd)

# Delete all endpoints
list_endpoints_cmd = "gcloud ai endpoints list --region=us-central1 --format='value(name)'"
endpoints = run_command(list_endpoints_cmd)

if endpoints:
    for endpoint in endpoints.strip().split('\n'):
        if endpoint:  # Check if endpoint is not empty
            delete_cmd = f"gcloud ai endpoints delete {endpoint} --region=us-central1 --quiet"
            print(f"Deleting endpoint: {endpoint}")
            run_command(delete_cmd)

print("Deletion process completed!")
```
